```python
columns_to_drop = ['Patient UHID', 'Height','Weight','HR','location','Risk Stratification',
       'Procedure', 'Image Reproduction', 'Findings', 'LM', 'LAD', 'LCX',
       'RCA', 'AM', 'PDA', 'PLV', 'Vascular Pattern', 'Output/Impression','Presenting Complaints']

df = df.drop(columns=columns_to_drop)
```

```python
df.columns
```

```python
df.head()
```

```python
df['Sex'] = df['Sex'].replace({'Male': 0, 'Female': 1})
```

```python
df.head()
```

```python
def has_consecutive_nans(row, n):
    return any(row.isna().rolling(window=n).sum() == n)

# Remove rows with more than 8 consecutive NaNs
df = df[~df.apply(lambda row: has_consecutive_nans(row, 8), axis=1)]
```

```python
df.head()
```

```python
df = df.fillna(0)
```

```python
df.head()
```

```python
df['Additional High Risk Symptom'] = df['Additional High Risk Symptom'].apply(lambda x: 1 if str(x).lower() == 'chest pain' else 0)
```
