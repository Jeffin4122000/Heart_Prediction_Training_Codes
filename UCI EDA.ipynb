```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=df1, x='age', hue='target', multiple='stack', edgecolor='black', alpha=0.7, bins=10)
plt.title('Age Distribution by Outcome')
plt.xlabel('Age')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=df1, x='sex', hue='target', multiple='stack', edgecolor='black', alpha=0.7, bins=2)
plt.title('sex vs o/p')
plt.xlabel('sex')
plt.ylabel('Output')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.kdeplot(data=df1, x='chol', hue='sex', fill=True, alpha=0.5)
plt.title('Density Plot of chol by Sex')
plt.xlabel('Binary Output')
plt.ylabel('Density')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=df1, x='age', hue='restecg', multiple='stack', edgecolor='black', alpha=0.7, bins=10)
plt.title('age vs ecg')
plt.xlabel('age')
plt.ylabel('rest ecg')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.kdeplot(data=df1, x='thal', hue='sex', multiple='stack', edgecolor='black')
plt.title('thalassemia vs sex')
plt.xlabel('Sex')
plt.ylabel('')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=df1, x='RF6(Obesity)', hue='Sex', multiple='stack', edgecolor='black', alpha=0.7, bins=2)
plt.title('Obesity vs Sex')
plt.xlabel('Obesity')
plt.ylabel(' ')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=df1, x='Calcification Score', hue='Sex', multiple='stack', edgecolor='black', alpha=1, bins=3)
plt.title('Calcification Score vs Sex')
plt.xlabel('Calcification Score')
plt.ylabel(' ')
plt.show()
```
